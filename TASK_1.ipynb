```python
Q1=x.quantile(0.25)
Q3=x.quantile(0.75)
IQR=Q3-Q1
```

```python
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
```

```python
mask=~((x<lower)|(x>upper)).any(axis=1)
x_clean=x[mask]
y_clean=y[mask]
```
